## 1. Numbers divisible by 7 but not by 3

**Question:** Find all numbers between 2000 and 3200 (inclusive) that are divisible by 7 but are not multiples of 3. Print them as a comma-separated sequence.

````python
numbers = []

for i in range(2000, 3201):
    if i % 7 == 0 and i % 3 != 0:
        numbers.append(str(i))

print(",".join(numbers))
````

## 2. Convert a list of characters into a string

**Question:** Convert the list `["C", "O", "D", "I", "N", "G"]` into the string `Coding`.

````python
characters = ["C", "O", "D", "I", "N", "G"]

result = "".join(characters).capitalize()

print(result)
````

## 3. Create a dictionary of numbers and their squares

**Question:** Given an integer `n`, create a dictionary containing each number `i` from 1 to `n` as a key and `i * i` as its value.

````python
n = int(input("Enter a number: "))

squares = {}

for i in range(1, n + 1):
    squares[i] = i * i

print(squares)
````

**Example output for input `5`:**
```text
{1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```

**Explanation:**
- `{}` creates an empty dictionary.
- `range(1, n + 1)` generates numbers from 1 to `n`.
- `squares[i] = i * i` stores each number and its square in the dictionary.
